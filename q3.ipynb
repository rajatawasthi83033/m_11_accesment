Predicting Diabetes Risk via REST API 

Problem:
Develop a REST API that predicts a person’s risk of diabetes based on health parameters such as BMI, age, blood pressure, and glucose levels. The API should accept JSON input and return the predicted risk score or classification.

Task:
1.	Train a classification model (Logistic Regression or Random Forest) on a diabetes dataset and save the model.
2.	Build a Flask or FastAPI application with a /predict endpoint.
3.	Accept JSON input for user health parameters and preprocess it for the model.
4.	Return the prediction (risk probability or class) in JSON format.
5.	Test the API locally and validate predictions using sample JSON requests.
